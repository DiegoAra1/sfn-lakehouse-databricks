```python
%sql
CREATE CATALOG IF NOT EXISTS sfn_dev
COMMENT 'Ambiente de desenvolvimento do projeto SFN Lakehouse';
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS sfn_dev.landing
COMMENT 'Landing zone para arquivos brutos recebidos das fontes';

CREATE SCHEMA IF NOT EXISTS sfn_dev.bronze
COMMENT 'Dados ingeridos preservando estrutura e conteúdo da origem';

CREATE SCHEMA IF NOT EXISTS sfn_dev.silver
COMMENT 'Dados tratados, tipados, deduplicados e validados';

CREATE SCHEMA IF NOT EXISTS sfn_dev.quarantine
COMMENT 'Registros rejeitados pelas regras de qualidade';

CREATE SCHEMA IF NOT EXISTS sfn_dev.gold
COMMENT 'Dados modelados para consumo analítico';

CREATE SCHEMA IF NOT EXISTS sfn_dev.ops
COMMENT 'Auditoria, métricas e observabilidade dos pipelines';
```

```python
%sql
SHOW SCHEMAS IN sfn_dev;
```
