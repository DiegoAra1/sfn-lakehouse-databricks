```python
import requests

BASE_URL = "https://olinda.bcb.gov.br/olinda/servico/IFDATA/versao/v1/odata"
ANO_MES = 202403
```

```python
url = (
    f"{BASE_URL}/IfDataCadastro(AnoMes=@AnoMes)"
    f"?@AnoMes={ANO_MES}&$top=5&$format=json"
)

response = requests.get(url, timeout=30)

print("Status:", response.status_code)
print("URL:", response.url)
print("Resposta:", response.text[:1000])
```

```python
test_url = (
    f"{BASE_URL}/ListaDeRelatorio()"
    "?$top=5&$format=json"
)

test_response = requests.get(
    test_url,
    timeout=30
)

print("Status:", test_response.status_code)
print("Resposta:", test_response.text[:1000])
```

```python
from datetime import datetime, timezone
from pathlib import Path
import json

# Momento da ingestão
ingestion_ts = datetime.now(timezone.utc)
ingestion_date = ingestion_ts.strftime("%Y-%m-%d")
timestamp_file = ingestion_ts.strftime("%Y%m%dT%H%M%SZ")

# Caminho da Landing/Raw
raw_dir = Path(
    f"/Volumes/sfn_dev/landing/raw_files/"
    f"bcb/ifdata/report_list/"
    f"ingestion_date={ingestion_date}"
)

raw_dir.mkdir(parents=True, exist_ok=True)

# Nome único do arquivo
raw_file = raw_dir / f"ifdata_report_list_{timestamp_file}.json"

# Salva exatamente a resposta recebida da API
with open(raw_file, "w", encoding="utf-8") as f:
    f.write(test_response.text)

print("Arquivo salvo em:")
print(raw_file)
```
